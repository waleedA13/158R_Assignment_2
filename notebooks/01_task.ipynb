## 1. The task

Given a user and an item in a specific size, plus the user's body measurements, we predict how the item fit: small, fit, or large. One rental transaction is one example. The data is the Rent the Runway fit dataset from Misra, Wan, and McAuley (RecSys 2018): 105,508 users, 5,850 items, and 192,544 transactions. This three-class problem fits the course: logistic regression on features, and latent factor models with user and item biases.

## 2. Evaluation

The primary metric is average AUC, taken over the three classes. Misra et al. report the same metric, so our numbers sit next to theirs. An AUC of 0.5 is random ranking and higher is better. Their paper reports 0.61 for its simplest baseline and 0.719 for its own model.

The secondary metric is macro-F1, which weights the three classes equally. About 74% of labels are "fit", so plain accuracy would reward a model that always says "fit". Macro-F1 does not.

## 3. Baselines

1. **Always predict "fit".** This is the floor under class imbalance. Any model has to beat it.
2. **Logistic regression on size and measurement features.** This is the course model. It sees the same inputs as our main model but has no user or item factors.

The main model adds user and item latent factors with biases on top of the features, with class weighting. Because baseline 2 shares the features, any gain of the main model over it comes from the latent factors and the weighting, not from extra inputs.

## 4. Validity

We split the transactions at random into 80% train, 10% validation, and 10% test, with seed 158, through one shared function. We choose models and settings on validation only. We score the test set once, at the end. The leakage guard is planned in general terms: the split is fixed by the seed, and no test statistic enters preprocessing or training.

To measure variance, we repeat the split and training over several seeds, or bootstrap the test predictions, and report intervals next to each score. Sanity checks: the "always fit" baseline should score an AUC of 0.5, and our scores should fall in a believable range against the paper's 0.61 to 0.719.
